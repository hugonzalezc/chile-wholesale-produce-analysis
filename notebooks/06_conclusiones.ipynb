## Lectura de conjunto

La pregunta general del proyecto era qué dicen estos datos sobre la comercialización mayorista de frutas y hortalizas en Chile. Con la salvedad de que solo cubren los 12 mercados que monitorea ODEPA, en 9 regiones:

- **Concentración.** La Región Metropolitana reúne entre 44% y 60% del volumen monitoreado, y Lo Valledor sola entre un cuarto y un tercio. No hay evidencia de que esa concentración haya aumentado o disminuido de forma clara entre 2020 y 2026: depende de si se mira al producto típico o al volumen total.
- **Origen.** Alrededor del 90% de las filas reportadas son de origen nacional. La participación extranjera sube un par de puntos y está concentrada en pocos productos que casi siempre se importan.
- **Precios.** Entre 2020 y 2025 el precio mayorista del producto mediano subió menos que la inflación general, con un cambio real de −7,1%. Hay productos que suben en términos reales (plátano y piña, entre otros) y otros que bajan con fuerza (papa, zapallo, palta). Ese resultado agregado se mantiene con otros años base, pero la clasificación por producto no es estable.
- **Mercados.** Hay mercados más y menos volátiles, pero la diferencia es moderada y no se asocia con el tamaño.
- **Calidad y temporada.** La calidad se paga (una prima típica de ~21% de Primera sobre Segunda, con mucha variación entre productos), y la mayoría de los productos evaluables tiene un patrón estacional estable, aunque no tomate ni cebolla.

**Lo que estos datos no dicen:** por qué ocurre nada de lo anterior. Son asociaciones y diferencias observadas. Las razones posibles (tipo de cambio, clima, estructura del mercado, cambios en cómo se reporta) son hipótesis que habría que contrastar con otros datos.

## Lo que el proyecto enseñó sobre estos datos

1. **La unidad de venta cambia.** Hay 191 etiquetas de unidad para 82 productos. Comparar precios o volúmenes sin controlar por la unidad es engañoso. Se resolvió comparando dentro de una misma etiqueta y validando el precio por kilo: con el mismo envase, la diferencia mediana entre etiquetas es de 6,1% (D-27).
2. **`Volumen` no tiene unidad documentada.** Se trabajó con una hipótesis no confirmada (D-14), y lo que depende de ella se presenta como sensibilidad, no como resultado principal.
3. **Dos medidas iniciales estaban sesgadas y se corrigieron.** La medida de P3 por etiqueta principal confundía envase con origen (D-21), y la estabilidad estacional de P6 con precios nominales se contaminaba con la inflación (D-31). En ambos casos el problema se detectó comparando contra otra medida o con datos sintéticos, no mirando solo el resultado.
4. **El IPC armado a mano mezclaba dos bases.** Sin empalmar, la inflación de 2020 a 2025 salía +3,2% en vez de +36,9% (D-32).
5. **Los datos traen anomalías que hay que declarar.** Mango tiene el mismo precio para Primera y Segunda en la mayoría de sus pares, los precios se cotizan en números redondos, la cobertura de algunos mercados cambia entre años (ajo en 2020) y 2026 está incompleto.

## Limitaciones generales

- Solo se cubren los 12 mercados que ODEPA monitorea, no todo el comercio del país.
- La unidad de `Volumen` no está confirmada por la fuente. Las medidas en kilos dependen de esa hipótesis.
- Los precios son nominales, salvo en P4. El IPC es el general de hogares urbanos, no un índice de alimentos, y se compara con precios mayoristas.
- Todos los resultados son asociaciones, no relaciones causales.
- P6 evalúa 44 de los 82 productos: las frutas de temporada corta y algunos productos con pocos mercados quedaron fuera.
- Varias comparaciones se apoyan en pocos mercados por producto (3 o 4 en 9 de los 32 productos de P4).
- La clasificación de P4 por producto depende del año base.
